```python
import datasets 

data = datasets.load_dataset("json", data_files="./data/dataset.json", split="train").shuffle(seed=42)
```

```python
data = data.class_encode_column("label")
split_data = data.train_test_split(test_size=0.2, stratify_by_column="label", seed=42)
```

```python
split_data.save_to_disk("./data/input_data")
```
